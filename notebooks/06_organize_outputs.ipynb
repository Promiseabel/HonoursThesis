```python
import json
import os
from collections import defaultdict
import numpy as np

METHODS = ["groundtruth", "mediapipe", "mmpose_halpe", "yolo_coco"]

FRAMES = {
    "close": [
        "maxV_close_frame_7",
        "maxV_close_frame_11",
        "maxV_close_frame_15"
    ],
    "far": [
        "maxV_far_frame_30",
        "maxV_far_frame_35",
        "maxV_far_frame_40"
    ],
    "accel": [
        "accel_frame_57",
        "accel_frame_59",
        "accel_frame_61",
        "accel_frame_67",
        "accel_frame_69"
    ],
    "maxV": [
        "maxV_close_frame_9",
        "maxV_close_frame_10",
        "maxV_close_frame_11",
        "maxV_close_frame_12",
        "maxV_close_frame_19"
    ],
    "all": [
        "maxV_close_frame_7",
        "maxV_close_frame_11",
        "maxV_close_frame_15",
        "maxV_far_frame_30",
        "maxV_far_frame_35",
        "maxV_far_frame_40",
        "accel_frame_57",
        "accel_frame_59",
        "accel_frame_61",
        "accel_frame_67",
        "accel_frame_69",
        "maxV_close_frame_9",
        "maxV_close_frame_10",
        "maxV_close_frame_11",
        "maxV_close_frame_12",
        "maxV_close_frame_19"
    ]
}

keypoints = ["keypoints"]
angles = ["angles"]
```

```python

# Function to load JSON data
def load_json(file_path):
    if os.path.exists(file_path):
        with open(file_path, "r") as f:
            return json.load(f)
    return []
```

```python
def concatenate_json(phase,pathsToCheck,file_type):
    base_dir = f"../comparison_final/{file_type[0]}/"
    output_dir = os.path.join(base_dir, phase)
    os.makedirs(output_dir, exist_ok=True)
    all_data = {method: {ftype: defaultdict(dict) for ftype in file_type} for method in METHODS}
    # Process each subdirectory
    for sub_dir in os.listdir(base_dir):
        sub_path = os.path.join(base_dir, sub_dir)

        # Skip any other directory
        if sub_dir not in pathsToCheck:
            continue

        # Process each method and file type
        for method in METHODS:
            for ftype in file_type:
                file_name = f"{method}_{ftype}.json"
                file_path = os.path.join(sub_path, file_name)

                # Load JSON data
                data = load_json(file_path)

                # Process frames and update names
                for entry in data:
                    new_frame_id = f"{sub_dir}_{entry['frame']}"  # Prefix with directory name
                    entry["frame"] = new_frame_id
                    if entry["frame"] in FRAMES[phase]:
                        all_data[method][ftype][new_frame_id] = entry  # Store using new frame ID

    # Save each combined file separately
    for method in METHODS:
        for ftype in file_type:
            output_file = os.path.join(output_dir, f"{method}_{ftype}.json")
            
            # Convert dict to list and save
            merged_list = list(all_data[method][ftype].values())
            with open(output_file, "w") as f:
                json.dump(merged_list, f, indent=4)

            print(f"Saved: {output_file}")
```

```python

concatenate_json("close",["maxV_close"],keypoints)

concatenate_json("far",["maxV_far"],keypoints)

concatenate_json("accel",["accel"],keypoints)

concatenate_json("maxV",["maxV_close"],keypoints)

concatenate_json("all",["accel","maxV_close","maxV_far"],keypoints)




concatenate_json("close",["maxV_close"],angles)

concatenate_json("far",["maxV_far"],angles)

concatenate_json("accel",["accel"],angles)

concatenate_json("maxV",["maxV_close"],angles)

concatenate_json("all",["accel","maxV_close","maxV_far"],angles)

```
