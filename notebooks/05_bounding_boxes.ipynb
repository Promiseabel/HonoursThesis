# This Notebook will figure out the size ratio of the sprinter in each frame accross all data

```python
import json

# Load JSON file
file= "maxV_close"
file_path = f"../FINAL/BBoxV2/{file}.json"  # Change this to your actual file path
with open(file_path, "r") as f:
    data = json.load(f)

# Process each frame to compute size ratio
frame_size_ratios = []
for frame in data:
    frame_name = frame["frame_image"].split("/")[-1].split(".")[0].split("-")[-1]
    original_width = frame["bounding_box"][0]["original_width"]
    original_height = frame["bounding_box"][0]["original_height"]
    
    # Compute frame area
    frame_area = original_width * original_height

    # Compute bounding box area
    bbox = frame["bounding_box"][0]
    actual_width = (bbox["width"] / 100.0) * original_width
    actual_height = (bbox["height"] / 100.0) * original_height
    bbox_area = actual_width * actual_height
    
    # Compute bounding box diagonal
    bbox_diagonal = (actual_width**2 + actual_height**2)**0.5

    # Compute bounding box area
    head_bbox = frame["head_bounding_box"][0]
    actual_width = (head_bbox["width"] / 100.0) * original_width
    actual_height = (head_bbox["height"] / 100.0) * original_height
    
    # Compute bounding box diagonal
    head_bbox_diagonal = (actual_width**2 + actual_height**2)**0.5

    normalized_head_bbox_diagonal = head_bbox_diagonal * 0.6

    # Compute size ratio as percentage
    size_ratio = (bbox_area / frame_area) * 100
    
    # Store results
    frame_size_ratios.append({"frame_name": frame_name, "size_ratio": size_ratio, "bbox_diagonal": bbox_diagonal, "normalized_head_bbox_diagonal": normalized_head_bbox_diagonal, "head_bbox_diagonal": head_bbox_diagonal})

# Save to JSON file
output_file = f"../FINAL/BBoxV2/{file}_size_ratios.json"  # Change this to your desired output path
with open(output_file, "w") as f:
    json.dump(frame_size_ratios, f, indent=4)

print(f"Size ratio data saved to {output_file}")
```
